```python
import pandas as pd
```

```python
df = pd.read_csv('MetObjects.txt', low_memory=False)
df
```

```python
df.columns.tolist()
```

```python
keywords = ['buddha', 'buddhist', 'bodhisattva']

mask = (
    df['Title'].str.contains('|'.join(keywords), case=False, na=False) |
    df['Object Name'].str.contains('|'.join(keywords), case=False, na=False) |
    df['Tags'].str.contains('|'.join(keywords), case=False, na=False)
)

buddha_df = df[mask]
```

```python
print(f"Found {len(buddha_df)} items related to Buddha/Buddhism.")
buddha_df[['Object Number', 'Title', 'Culture', 'Object Name', 'Classification', 'Link Resource']].head(10)
```

```python
buddha_sculptureCC0 = buddha_df[(buddha_df['Is Public Domain'] == True) & (buddha_df['Classification'].str.contains('Sculpture', case=False, na=False))]
buddha_sculptureCC0
```

```python
buddha_sculptureCC0 = buddha_sculptureCC0[['Object ID', 'Object Name', 'Title', 'Culture', 'Medium', 'Link Resource']]
buddha_sculptureCC0
```

```python
buddha_sculptureCC0['Object Name'].unique()
```

```python
buddha_sculptureCC0['Object Name'] = buddha_sculptureCC0['Object Name'].str.lower().str.replace('\r\n', '', regex=False)
buddha_sculptureCC0
```

```python
exclude_names = ['Portable shrine', 'Stupa', 'Head', 'Head\r\n', 'Bust', 'Bust fragment', 'Relief fragment', 'Architectural element', 'Relief fragment','Fragment','Fragment of halo','Pagoda base','Drum slab','Stupa model','Base of stele','Upper half of figure','Part of figure','Upper half of statue','Reliquary', 'Reliquary (?)', 'Mold', 'Mold and impression']
buddha_sculptureCC0 = buddha_sculptureCC0[~buddha_sculptureCC0['Object Name'].isin(exclude_names)]
buddha_sculptureCC0
```

```python
buddha_sculptureCC0['Link Image'] = buddha_sculptureCC0['Object ID'].apply(lambda x: f"https://collectionapi.metmuseum.org/api/collection/v1/iiif/{x}/main-image")
buddha_sculptureCC0
```

```python
buddha_sculptureCC0.loc[buddha_sculptureCC0['Object ID'] == 39184, 'Link Image'] = "https://collectionapi.metmuseum.org/api/collection/v1/iiif/39184/restricted"
buddha_sculptureCC0.loc[buddha_sculptureCC0['Object ID'] == 50668, 'Link Image'] = "https://collectionapi.metmuseum.org/api/collection/v1/iiif/50668/restricted"
buddha_sculptureCC0
```

```python
keywords_to_filter = ['figure', 'figures', 'figurine', 'headless figure']
buddha_sculptureCC0_fig = buddha_sculptureCC0[
    buddha_sculptureCC0['Object Name'].str.contains('|'.join(keywords_to_filter), case=False, na=False)
]
buddha_sculptureCC0_fig
```

```python
buddha_sculptureCC0_fig.to_csv('buddha_sculptureCC0_fig.csv', index=False)
```
